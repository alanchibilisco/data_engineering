## P1-E3:
¿Cuál es el viaje más largo (mayor distancia) en el dataset? ¿Y el más corto (excluyendo
ceros)?

```sql
%sql

SELECT 
MAX(trip_distance) as distancia_maxima,
MIN(trip_distance) as distancia_minima,
distancia_maxima - distancia_minima as diferencia
--MIN(trip_distance) as distancia_minima 
FROM samples.nyctaxi.trips
WHERE trip_distance >0;
    
SELECT 
MAX(trip_distance) as distancia_maxima,
MIN(trip_distance) as distancia_minima,
MAX(trip_distance) - MIN(trip_distance) as diferencia
--MIN(trip_distance) as distancia_minima 
FROM samples.nyctaxi.trips
WHERE trip_distance >0;

```

## P1-E4:
¿Cuál es el promedio de tarifa (fare_amount) y distancia (trip_distance) de los viajes?

```sql
%sql

SELECT
ROUND(AVG(fare_amount),2) as tarifa_promedio,
ROUND(AVG(trip_distance),2) as distancia_promedia
FROM trips;
```

## P1-E5:
Muestra los 20 viajes más recientes con su fecha, distancia y tarifa.

```sql
%sql

SELECT tpep_pickup_datetime as fecha, trip_distance as distancia, fare_amount as tarifa
FROM trips
ORDER BY tpep_pickup_datetime DESC
LIMIT 20;
```

```sql
%sql
SELECT
ROUND(AVG(fare_amount),2) as tarifa_promedio,
ROUND(AVG(trip_distance),2) as distancia_promedia
FROM trips
WHERE fare_amount > 0 AND trip_distance >0;
```

## P6-E6.1
Para zonas con más de 1,000 viajes, muestra: zona, cantidad de viajes, tarifa máxima, promedio y mínima. Solo incluye viajes con tarifa mayor a $5 y distancia mayor a 0.

```sql
%sql

select * from trips limit 1;

select
pickup_zip as zona,
count(*) as cantidad_viajes,
round(max(fare_amount),2) as tarifa_max,
round(avg(fare_amount),2) as tarifa_avg,
round(min(fare_amount),2) as tarifa_min
from trips
where fare_amount > 5 AND trip_distance > 0
group by pickup_zip
having count(*) > 1000;
```

## P6-E6.2

```sql
select * from trips limit 1;


select 
    case 
    when trip_distance < 2 then "Corto"
    when trip_distance between 2 and 5 then "Medio"
    when trip_distance between 5 and 10 then "Largo"
    else "Muy largo"
    end as categoria,
    count(*) as cantidad_viajes,
    round(avg(fare_amount),2) as tarifa_avg,
    round(avg(trip_distance),2) as distancia_avg
from
trips
where trip_distance > 0 and fare_amount > 0
group by categoria
having count(*) > 100
order by tarifa_avg desc;

SELECT
  CASE
    WHEN trip_distance < 2 THEN 'Corto (< 2 mi)'
    WHEN trip_distance BETWEEN 2 AND 5 THEN 'Medio (2-5 mi)'
    WHEN trip_distance BETWEEN 5 AND 10 THEN 'Largo (5-10 mi)'
    ELSE 'Muy largo (> 10 mi)'
  END as categoria,
  COUNT(*) as total_viajes,
  ROUND(AVG(fare_amount), 2) as tarifa_promedio,
  ROUND(AVG(trip_distance), 2) as distancia_promedio
FROM samples.nyctaxi.trips
WHERE trip_distance > 0
  AND fare_amount > 0
GROUP BY
  CASE
    WHEN trip_distance < 2 THEN 'Corto (< 2 mi)'
    WHEN trip_distance BETWEEN 2 AND 5 THEN 'Medio (2-5 mi)'
    WHEN trip_distance BETWEEN 5 AND 10 THEN 'Largo (5-10 mi)'
    ELSE 'Muy largo (> 10 mi)'
  END
HAVING COUNT(*) > 100
ORDER BY tarifa_promedio DESC;
```
