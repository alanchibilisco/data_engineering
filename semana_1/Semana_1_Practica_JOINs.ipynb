# Parte 2 INNER JOIN

## P2-E2.1

```sql
%sql

select
    c.cliente_id,
    c.nombre,
    c.ciudad,
    o.orden_id,
    o.producto_id,
    o.cantidad,
    o.fecha
from bootcamp.practice.clientes c
inner join bootcamp.practice.ordenes o ON o.cliente_id = c.cliente_id;
```

## P2-E2.2

```sql
%sql

select
    o.orden_id,
    o.fecha,
    p.nombre,
    p.categoria,
    o.cantidad,
    p.precio,
    o.cantidad * p.precio as total
from bootcamp.practice.ordenes o
inner join bootcamp.practice.productos p ON o.producto_id = p.producto_id
order by o.orden_id;   
```

## P2-E2.3

```sql
%sql

select
    c.nombre as cliente,
    c.ciudad,
    o.orden_id,
    o.fecha,
    p.nombre as producto,
    o.cantidad,
    p.precio as precio_unitario,
    p.precio * o.cantidad as total
from bootcamp.practice.ordenes as o
inner join bootcamp.practice.clientes as c ON o.cliente_id = c.cliente_id
inner join bootcamp.practice.productos as p ON o.producto_id = p.producto_id
order by o.fecha desc;
```

# PARTE 3 LEFT JOIN

## P3-E3.1

```sql
%sql

select
 *
from bootcamp.practice.clientes c 
left join bootcamp.practice.ordenes o on o.cliente_id=c.cliente_id
order by c.cliente_id;

select
c.cliente_id,
c.nombre,
COUNT(*) as total_ordenes
from bootcamp.practice.clientes c 
left join bootcamp.practice.ordenes o on o.cliente_id=c.cliente_id
group by c.cliente_id, c.nombre
having total_ordenes<2
order by c.cliente_id;
```

## P3-E3.2

```sql
%sql

select
c.cliente_id,
c.nombre,
o.orden_id
from bootcamp.practice.clientes c
left join bootcamp.practice.ordenes o on o.cliente_id=c.cliente_id
where o.orden_id is null
order by c.cliente_id;
```

# PARTE 7

## P7-E7.1

```sql
%sql

select
    c.nombre as cliente,
    c.ciudad,
    o.orden_id,
    o.fecha,
    p.nombre as producto,
    p.categoria,
    o.cantidad,
    p.precio as p_unit,
    p.precio * o.cantidad as p_total
from bootcamp.practice.clientes as c
inner join bootcamp.practice.ordenes as o on o.cliente_id=c.cliente_id
inner join bootcamp.practice.productos as p on p.producto_id=o.producto_id
order by c.nombre, o.fecha;
```

## P7-E7.2

```sql
%sql

select
    c.nombre as cliente,
    c.ciudad,
    COUNT(o.*) as total_ordenes,
     COALESCE(SUM(o.cantidad * p.precio), 0) as monto_total_gastado
from bootcamp.practice.clientes as c
left join bootcamp.practice.ordenes as o on o.cliente_id=c.cliente_id
left join bootcamp.practice.productos as p on p.producto_id=o.producto_id
group by c.cliente_id, c.nombre, c.ciudad
order by c.nombre;
```

## P7-E7.3

```sql
%sql

select
 p.nombre as producto,
 p.categoria,
 p.precio as precio_unit,
 coalesce(sum(o.cantidad),0) as cantidad_vendida,
 coalesce(sum(o.cantidad*p.precio),0) as ingresos_totales
 from bootcamp.practice.productos as p
 left join bootcamp.practice.ordenes as o on o.producto_id=p.producto_id
 group by p.producto_id, p.nombre, p.categoria, p.precio
 order by p.categoria, p.nombre;
    
```
